# Analyse des valeurs aberrantes (Outliers)

```python
# Calcul du nombre de valeurs aberrantes pour chaque variable avec la méthode IQR

for colonne in df.select_dtypes(include=np.number).columns:

    Q1 = df[colonne].quantile(0.25)
    Q3 = df[colonne].quantile(0.75)

    IQR = Q3 - Q1

    borne_inf = Q1 - 1.5 * IQR
    borne_sup = Q3 + 1.5 * IQR

    nombre_outliers = ((df[colonne] < borne_inf) | 
                       (df[colonne] > borne_sup)).sum()

    print(colonne, ":", nombre_outliers)
```

```python
# Création des Boxplots pour chaque variable numérique

colonnes_numeriques = df.select_dtypes(include=np.number).columns

fig, axes = plt.subplots(3, 3, figsize=(16, 12))

axes = axes.flatten()

for i, colonne in enumerate(colonnes_numeriques):
    sns.boxplot(y=df[colonne], ax=axes[i])
    axes[i].set_title(colonne, fontsize=9)
    axes[i].set_ylabel("Valeurs")

plt.suptitle("Analyse des valeurs aberrantes par variable", fontsize=15)

plt.tight_layout()
plt.show()
```

```python
### Conclusion :

#L'analyse des valeurs aberrantes avec la méthode IQR a permis de détecter plusieurs valeurs extrêmes dans certaines variables,
#notamment Age, Water et Superplasticizer.

#Cependant, ces valeurs ne représentent pas nécessairement des erreurs.
#Certaines peuvent correspondre à des observations expérimentales normales.

#Par conséquent, nous avons choisi de conserver les valeurs aberrantes pour le moment,
#afin de ne pas supprimer des informations potentiellement importantes pour la prédiction de la résistance du béton.
```

# Définition des variables explicatives (X) et de la variable cible (y)

```python
# Définition des variables explicatives
X = df.drop("Concrete compressive strength(MPa, megapascals)", axis=1)

# Définition de la variable cible
y = df["Concrete compressive strength(MPa, megapascals)"]
```

```python
print("Dimensions de X :", X.shape)
print("Dimensions de y :", y.shape)

#Après cette étape, nous disposons de 1005 observations, de 8 variables explicatives (X) et d'une (1) variable cible (y)( Concrete compressive strength(MPa, megapascals)). 
#Les données sont maintenant organisées pour passer à l'étape suivante : 
#la séparation en ensembles d'entraînement et de test (Train/Test Split).
```

# Application du découpage Train/Test

```python
#La séparation Train/Test permet donc de distinguer les données utilisées pour l'apprentissage de celles utilisées pour l'évaluation.
#Dans notre projet, nous utiliserons 80% des observations pour l'entraînement et 20% pour le test.
```

```python
# Importation de la fonction train_test_split

from sklearn.model_selection import train_test_split
```

```python
# Division des données en ensembles d'entraînement et de test

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,   # 20% pour le test et 80% pour l'entraînement
    random_state=42  # pour obtenir le même découpage à chaque exécution
)
```

```python
# Affichage des dimensions des ensembles Train et Test

print("Dimensions de X_train :", X_train.shape)
print("Dimensions de X_test :", X_test.shape)

print("Dimensions de y_train :", y_train.shape)
print("Dimensions de y_test :", y_test.shape)
```

# Identification des groupes (Groupes de compositions)

```python
# Sélection des 7 variables représentant la composition du béton

colonnes_composition = X.columns[:7]

# Création d'un groupe pour chaque composition identique

groupes = df[colonnes_composition].apply(tuple, axis=1)
```

```python
# Calcul du nombre de groupes différents

print("Nombre total d'observations :", len(groupes))
print("Nombre de groupes différents :", groupes.nunique())
```

```python

#Après l'identification des groupes, nous avons obtenu 1005 observations et 428 groupes de compositions différentes.

#Cela signifie que plusieurs observations possèdent les mêmes quantités des sept composants du béton, mais peuvent avoir des âges différents.

#Chaque groupe représente donc une composition unique du béton, indépendamment de son âge.

#Cette identification est nécessaire pour appliquer le Grouped Split. 
#Elle permet de garantir que les observations appartenant à une même composition ne seront pas réparties entre les ensembles d'entraînement et de test.
```

```python
# Importation de la méthode GroupShuffleSplit pour diviser les groupes
from sklearn.model_selection import GroupShuffleSplit
```

```python
# Attribution d'un identifiant numérique à chaque composition
groupes_id = pd.factorize(groupes)[0]

# Vérification du nombre de groupes différents
print("Nombre de groupes différents :", len(np.unique(groupes_id)))
```

```python

# Création d'un objet GroupShuffleSplit
gss = GroupShuffleSplit(
    n_splits=1,       # Réaliser une seule division des données
    test_size=0.20,   # Réserver environ 20 % des groupes pour le test
    random_state=42   # Fixer la graine aléatoire pour obtenir le même découpage
)

# Division des données en ensembles d'entraînement et de test
# La division respecte les groupes de compositions du béton
train_idx, test_idx = next(gss.split(X, y, groups=groupes_id))
```

```python
# Création de l'ensemble d'entraînement (Train)
X_train_group = X.iloc[train_idx]  # Sélection des variables explicatives pour l'entraînement
y_train_group = y.iloc[train_idx]  # Sélection des valeurs cibles correspondantes

# Création de l'ensemble de test (Test)
X_test_group = X.iloc[test_idx]    # Sélection des variables explicatives pour le test
y_test_group = y.iloc[test_idx]    # Sélection des valeurs cibles correspondantes
```

```python
# Affichage des dimensions de l'ensemble d'entraînement après la division par groupes.
print("Dimensions de X_train_group :", X_train_group.shape)
print("Dimensions de y_train_group :", y_train_group.shape)

# Affichage des dimensions de l'ensemble de test après la division par groupes.
print("Dimensions de X_test_group :", X_test_group.shape)
print("Dimensions de y_test_group :", y_test_group.shape)
```

```python
#Après la division par groupes, l'ensemble d'entraînement contient 799 observations et 8 variables explicatives, 
#tandis que l'ensemble de test contient 206 observations et 8 variables explicatives. 
#La répartition du nombre d'observations peut être différente de 80 % et 20 %,
#car GroupShuffleSplit effectue la division en fonction des groupes de compositions et non du nombre exact de lignes.
```

```python
# Récupération des groupes présents dans chaque ensemble
groupes_train = groupes_id[train_idx]
groupes_test = groupes_id[test_idx]

# Vérification des groupes communs entre Train et Test
groupes_communs = set(groupes_train) & set(groupes_test)

# Affichage du nombre de groupes communs
print("Nombre de groupes communs :", len(groupes_communs))
```

```python
#La vérification montre qu'il n'existe aucun groupe commun entre les ensembles d'entraînement et de test (0 groupe commun).
#Ainsi, toutes les observations appartenant à une même composition du béton sont conservées dans un seul ensemble.
#Cette séparation permet d'évaluer le modèle sur des compositions qui ne sont pas présentes dans les données d'entraînement.
```

```python
# Affichage des dimensions du Random Split
print("----- Random Split -----")
print("Train :", X_train.shape)
print("Test  :", X_test.shape)

# Affichage des dimensions du Grouped Split
print("\n----- Grouped Split -----")
print("Train :", X_train_group.shape)
print("Test  :", X_test_group.shape)
```

```python
#La différence entre les dimensions des deux méthodes est normale.
#Random Split divise les observations, tandis que Grouped Split divise les groupes de compositions. 
#Cette deuxième méthode garantit qu'une même composition ne se retrouve pas simultanément dans les ensembles d'entraînement et de test.
```

# Standardisation des données –Random Split

```python
# Importation de StandardScaler pour standardiser les variables(mise à l’échelle (Scaling))
from sklearn.preprocessing import StandardScaler

# Importation de Pipeline pour organiser le prétraitement et la modélisation
from sklearn.pipeline import Pipeline
```

```python
# Vérification des dimensions des données du Random Split
print("Dimensions de X_train :", X_train.shape)
print("Dimensions de X_test :", X_test.shape)
```

```python
# Création d'un objet StandardScaler
scaler = StandardScaler()

# Apprentissage des paramètres de standardisation sur les données d'entraînement
scaler.fit(X_train)

# Transformation des données d'entraînement
X_train_scaled = scaler.transform(X_train)

# Transformation des données de test avec les mêmes paramètres
X_test_scaled = scaler.transform(X_test)
```

```python
#La standardisation permet de mettre les variables explicatives sur une échelle comparable.
#Nous avons utilisé `fit()` sur les données d'entraînement pour calculer la moyenne et l'écart-type de chaque variable. 
#Ensuite, `transform()` a permis de transformer les données d'entraînement et de test avec les mêmes paramètres.
#Cette méthode évite d'utiliser les informations du test pendant le prétraitement et permet de limiter le risque de Data Leakage.
```

```python
# Affichage des noms des 8 variables explicatives
print("Noms des variables standardisées :")
print(X_train.columns.tolist())

# Calcul et affichage des moyennes de chaque variable après standardisation
print("\nMoyennes après standardisation :")
print(X_train_scaled.mean(axis=0))

# Calcul et affichage des écarts-types de chaque variable après standardisation
print("\nÉcarts-types après standardisation :")
print(X_train_scaled.std(axis=0))
```

```python
#Remarque sur le paramètre axis :

#Le paramètre `axis=0` permet d'effectuer les calculs colonne par colonne. Dans notre cas, 
#il permet de calculer la moyenne et l'écart-type séparément pour chacune des 8 variables explicatives.

#En revanche, `axis=1` permet d'effectuer les calculs ligne par ligne.
```

```python
#Après la standardisation, les moyennes des 8 variables explicatives sont très proches de 0 et leurs écarts-types sont égaux à 1.
#Ces résultats confirment que la standardisation a été effectuée correctement avec StandardScaler. 
#Les variables sont maintenant sur une échelle comparable.
```

```python
# Vérification des dimensions après standardisation
print("Dimensions de X_train_scaled :", X_train_scaled.shape)
print("Dimensions de X_test_scaled :", X_test_scaled.shape)
```

```python
#Après la standardisation, les dimensions des ensembles d'entraînement et de test restent inchangées.
#X_train_scaled contient toujours 804 observations et 8 variables explicatives, 
#tandis que X_test_scaled contient 201 observations et 8 variables explicatives.
#La standardisation modifie les valeurs des variables, mais ne modifie pas la structure des données.
```

# Standardisation des données – Grouped Split

```python
# Création d'un nouveau StandardScaler pour le Grouped Split
scaler_group = StandardScaler()

# Apprentissage des paramètres sur les données d'entraînement
scaler_group.fit(X_train_group)

# Standardisation des données d'entraînement
X_train_group_scaled = scaler_group.transform(X_train_group)

# Standardisation des données de test avec les mêmes paramètres
X_test_group_scaled = scaler_group.transform(X_test_group)
```

```python
# Affichage des noms des 8 variables explicatives
print("Noms des variables standardisées :")
print(X_train_group.columns.tolist())

# Vérification des moyennes des 8 variables après standardisation
# Les moyennes doivent être proches de zéro
print("\nMoyennes après standardisation :")
print(X_train_group_scaled.mean(axis=0))

# Vérification des écarts-types des 8 variables après standardisation
# Les écarts-types doivent être proches de 1
print("\nÉcarts-types après standardisation :")
print(X_train_group_scaled.std(axis=0))
```

```python
#Après la standardisation des données du Grouped Split, 
#les moyennes des 8 variables explicatives sont très proches de zéro et leurs écarts-types sont égaux à 1.
#Ces résultats montrent que la standardisation a été correctement appliquée aux données d'entraînement. 
#Les 8 variables sont maintenant sur une échelle comparable.
```

```python
# Affichage des dimensions des données d'entraînement après standardisation
print("Dimensions de X_train_group_scaled :")
print(X_train_group_scaled.shape)

# Affichage des dimensions des données de test après standardisation
print("\nDimensions de X_test_group_scaled :")
print(X_test_group_scaled.shape)
```

```python
#Après la standardisation, les dimensions des données restent inchangées.
#L'ensemble d'entraînement contient toujours 799 observations et 8 variables explicatives, 
#tandis que l'ensemble de test contient 206 observations et 8 variables explicatives.

#La standardisation a uniquement modifié les valeurs des variables, sans modifier la structure des données.
```
